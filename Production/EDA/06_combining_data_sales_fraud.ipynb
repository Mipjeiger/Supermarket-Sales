```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
from pathlib import Path
```

```python
pd.set_option('display.max_columns', None)
SALES_DIR = Path.cwd().parent / "app" / "data" / "cleaned" / "sql_supermarket.parquet"
FRAUD_DIR = Path.cwd().parent / "app" / "data" / "cleaned" / "fraud_sql_supermarket.parquet"
```

```python
df_sales = pd.read_parquet(SALES_DIR)
df_sales.head(3)
```

```python
df_fraud = pd.read_parquet(FRAUD_DIR)
df_fraud.head(3)
```

```python
df_fraud.info()
```

## Combining data

```python
df_combined = df_sales.merge(
    df_fraud[['order_id', 'shipping_days', 'fraud']],
    on='order_id',
    how='left'
)
df_combined.head()
```
