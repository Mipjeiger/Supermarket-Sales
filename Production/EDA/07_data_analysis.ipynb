```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

## Load database

```python
pd.set_option('display.max_columns', None)
df = pd.read_parquet("../../Production/app/data/cleaned/combined_sql_supermarket.parquet")
df.head()
```

## Data visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Build graph to find revenue by product_name
revenue_by_product = df.groupby('product_name')['sales'].sum().sort_values(ascending=False).head(10)

plt.Figure(figsize=(12, 8))
plt.barh(revenue_by_product.index, revenue_by_product.values)
plt.title('Top 10 Products by Revenue')
plt.xlabel('Revenue')
plt.ylabel('Product Name')
plt.show()
```
